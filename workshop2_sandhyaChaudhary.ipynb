2. Check info of the DataFrame and identify following:
(a) columns with dtypes=object
(b) unique values of those columns.
(c) check for the total number of null values in each column.

```python
import pandas as pd

df_readcsv = pd.read_csv('/content/drive/MyDrive/AI/Copy of bank.csv')
df = df_readcsv

print(" DataFrame Info ")
print(df.info())

# (a)
object_cols = df.select_dtypes(include="object").columns
print("\n(a) Columns with dtype = object:")
print(object_cols)

# (b)
print("\n(b) Unique values of object columns:")
for col in object_cols:
    print(f"\nColumn: {col}")
    print(df[col].unique())

# (c)
print("\n(c) Total number of null values in each column:")
print(df.isnull().sum())
```

3. Drop all the columns with dtypes object and store in new DataFrame, also write the DataFrame in
”.csv” with name ”banknumericdata.csv”

```python
# Drop all columns with dtype = object
df_numeric = df.select_dtypes(exclude="object")

# Display the new DataFrame
print(df_numeric.head())

# Save to CSV
df_numeric.to_csv("banknumericdata.csv", index=False)

print("banknumericdata.csv created successfully!")
```

2. Check info of the DataFrame and identify column with missing (null) values.

```python
import pandas as pd

df_med = pd.read_csv('/content/drive/MyDrive/AI/medical_students_dataset.csv')

# Check general info
print("DataFrame Info")
print(df_med.info())

# Check total null values in each column
print("\n Null Values in Each Column")
null_counts = df_med.isnull().sum()
print(null_counts)

# Show only columns that have missing values
print("\n Columns With Missing (Null) Values ")
missing_cols = null_counts[null_counts > 0]
print(missing_cols)
```

4. Check for any duplicate values present in Dataset and do necessary to manage the duplicate items.
{Hint: dataset.duplicated.sum()}

```python
# Check for duplicate rows
duplicate_count = df_med.duplicated().sum()
print(f"Total duplicate rows in dataset: {duplicate_count}")
# Display the duplicate rows
duplicates = df_med[df_med.duplicated()]
print("Duplicate rows:", duplicates)
# Remove duplicates and keep the first occurrence
df_med = df_med.drop_duplicates(keep='first')

# Verify duplicates removed
print(f"Duplicate rows after removal: {df_med.duplicated().sum()}")
```
